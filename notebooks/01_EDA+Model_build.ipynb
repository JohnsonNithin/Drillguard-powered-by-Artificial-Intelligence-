| Feature                | Simple meaning                                          | Why it matters                                                                              |
| ---------------------- | ------------------------------------------------------- | ------------------------------------------------------------------------------------------- |
| `Depth(m)`             | How far the well has been drilled                       | Rock and drilling conditions change with depth                                              |
| `weight on bit (kg)`   | How hard the drill bit is being pushed against the rock | More push may increase drilling speed, but too much can cause poor performance or vibration |
| `Rotary Speed (rpm)`   | How fast the drillstring or bit rotates                 | Faster rotation can increase ROP, but excessive speed can create wear or vibration          |
| `Surface Torque (KPa)` | How much effort is required to rotate the drillstring   | Increasing torque can suggest harder rock, friction, poor hole cleaning or sticking         |
| `Hookload (kg)`        | The weight supported by the rig’s lifting hook          | Helps indicate drillstring weight, contact with the bottom and friction inside the well     |
| `Pump Press (KPa)`     | Pressure used to pump drilling mud through the system   | Changes can indicate changing flow resistance, restrictions or operating conditions         |
| `Flow In (liters/min)` | Amount of drilling mud pumped into the well             | Mud cools the bit and carries broken rock upward                                            |
| `Flow Out %`           | Relative amount of mud returning from the well          | Unexpected changes may indicate fluid loss, gain or measurement problems                    |

# 7. Model Build

Inputs X:
Yeo–Johnson transformation
        ↓
MinMax scaling
        ↓
Linear Regression

Target y:
Yeo–Johnson transformation only
        ↓
Model
        ↓
Reverse transformation
        ↓
ROP in original units

```python
from sklearn.compose import TransformedTargetRegressor
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import MinMaxScaler, PowerTransformer

input_pipeline = Pipeline([
    (
        "yeo_johnson",
        PowerTransformer(
            method="yeo-johnson",
            standardize=False
        )
    ),
    ("minmax", MinMaxScaler())
])

linear_regressor = Pipeline([
    ("input_preparation", input_pipeline),
    ("model", LinearRegression())
])

linear_transformed_model = TransformedTargetRegressor(
    regressor=linear_regressor,
    transformer=PowerTransformer(
        method="yeo-johnson",
        standardize=False
    )
)
```

# 8. Model fit and prediction on training data snd testing data 

```python
linear_transformed_model.fit(X_train, y_train)

y_train_pred = linear_transformed_model.predict(X_train)
y_test_pred = linear_transformed_model.predict(X_test)
```

```python
import numpy as np
import pandas as pd

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
)

def calculate_metrics(y_true, y_pred):
    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R2": r2_score(y_true, y_pred),
    }
```

```python
results = pd.DataFrame({
    "Training": calculate_metrics(
        y_train,
        y_train_pred
    ),
    "Testing": calculate_metrics(
        y_test,
        y_test_pred
    ),
}).T

results
```

```python
from sklearn.ensemble import (
    ExtraTreesRegressor,
    GradientBoostingRegressor,
    RandomForestRegressor,
)

models = {
    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ),

    "Extra Trees": ExtraTreesRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        random_state=42
    ),
}
```

```python
model_results = []
fitted_models = {}
model_predictions = {}

for model_name, model in models.items():

    model.fit(X_train, y_train)

    train_predictions = model.predict(X_train)
    validation_predictions = model.predict(X_test)

    fitted_models[model_name] = model
    model_predictions[model_name] = validation_predictions

    train_metrics = calculate_metrics(
        y_train,
        train_predictions
    )

    validation_metrics = calculate_metrics(
        y_test,
        validation_predictions
    )

    model_results.append({
        "Model": model_name,
        "Dataset": "Training",
        **train_metrics
    })

    model_results.append({
        "Model": model_name,
        "Dataset": "Validation",
        **validation_metrics
    })
```

```python
tree_results = (
    pd.DataFrame(model_results)
    .set_index(["Model", "Dataset"])
    .round(3)
)

tree_results
```

```python
linear_results = pd.DataFrame([
    {
        "Model": "Transformed Linear Regression",
        "Dataset": "Training",
        **calculate_metrics(y_train, y_train_pred)
    },
    {
        "Model": "Transformed Linear Regression",
        "Dataset": "Validation",
        **calculate_metrics(y_test, y_test_pred)
    }
]).set_index(["Model", "Dataset"])

all_results = pd.concat([
    linear_results,
    tree_results
]).round(3)

all_results
```

# 11. Cross validation scores 

```python
from sklearn.ensemble import ExtraTreesRegressor, RandomForestRegressor
from sklearn.model_selection import KFold, cross_validate

cv_models = {
    "Random Forest": RandomForestRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=1
    ),
    "Extra Trees": ExtraTreesRegressor(
        n_estimators=300,
        random_state=42,
        n_jobs=1
    ),
}

cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scoring = {
    "mae": "neg_mean_absolute_error",
    "rmse": "neg_root_mean_squared_error",
    "r2": "r2",
}
```

```python
cv_results = []

for model_name, model in cv_models.items():

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring=scoring,
        n_jobs=-1
    )

    cv_results.append({
        "Model": model_name,

        "Mean MAE": -scores["test_mae"].mean(),
        "MAE variation": scores["test_mae"].std(),

        "Mean RMSE": -scores["test_rmse"].mean(),
        "RMSE variation": scores["test_rmse"].std(),

        "Mean R2": scores["test_r2"].mean(),
        "R2 variation": scores["test_r2"].std(),
    })
```

```python
cv_summary = (
    pd.DataFrame(cv_results)
    .set_index("Model")
    .round(3)
)

cv_summary
```

# Final Conclusion 

Random Forest was selected as the final Phase 1 model because it provided strong validation performance and stable cross-validation results. It predicts the dominant drilling conditions well but produces larger errors for rare, extreme ROP observations, likely because these conditions are underrepresented in the dataset, giving the model fewer examples from which to learn.

Also, Random Forest tends to average predictions, so it often struggles with rare extreme target values.
